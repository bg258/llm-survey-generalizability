## **1. Project and Meeting Objective**

The thesis looks at the extent to which LLM survey responses can be generalised across different questionnaires and data sources.

The aim of this meeting is to agree on a research question that is tractable and a first pilot study.

I'd be very interested in your thoughts on the project scope, the experimental methodology and the first deliverable.


## **2. Preliminary Research Question**

> How consistent are conclusions about one LLM across two questionnaires designed to measure the same construct?

A construct is the concept that the questionnaire is intended to measure (e.g. a particular value or attitude).

The initial idea is to pick two questionnaires that are measuring the same construct and see if they lead to the same conclusions about the model.

The construct and the questionnaires are yet to be selected with the supervisors.
